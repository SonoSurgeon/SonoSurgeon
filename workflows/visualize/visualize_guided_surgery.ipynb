```python
def plot_radial_vs_pos_with_time_color(pos_along_traj, radial_dist, cmap="viridis", linewidth=2, alpha=0.6):
    """
    Plots radial distance vs position along trajectory with color representing time progression.
    
    Parameters:
    - pos_along_traj: (N, T) tensor or array, position along trajectory
    - radial_dist: (N, T) tensor or array, radial distance from trajectory
    - cmap: str, colormap name (default: 'viridis')
    - linewidth: float, line width
    - alpha: float, line transparency
    """

    # Convert to numpy
    pos_along_traj = pos_along_traj.cpu().numpy() if torch.is_tensor(pos_along_traj) else pos_along_traj
    radial_dist = radial_dist.cpu().numpy() if torch.is_tensor(radial_dist) else radial_dist

    N, T = pos_along_traj.shape
    time = np.linspace(0, 1, T)  # normalized time for colormap

    fig, ax = plt.subplots(figsize=(7, 5))
    
    for i in range(N):
        x = pos_along_traj[i]
        y = radial_dist[i]

        # Create segments
        points = np.array([x, y]).T.reshape(-1, 1, 2)
        segments = np.concatenate([points[:-1], points[1:]], axis=1)

        # Map colors to time progression
        colors = cm.get_cmap(cmap)(time)

        lc = LineCollection(segments, colors=colors, linewidth=linewidth, alpha=alpha)
        ax.add_collection(lc)

    ax.set_xlabel("Position along trajectory", fontsize=12)
    ax.set_ylabel("Radial distance", fontsize=12)
    ax.set_title("Radial distance vs position along trajectory\n(color = time progression)", fontsize=14)

    # Add colorbar for time progression
    norm = plt.Normalize(0, 1)
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = fig.colorbar(sm, ax=ax)
    cbar.set_label("Normalized time", fontsize=12)

    ax.autoscale()
    ax.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()

    
```

```python
plot_radial_vs_pos_with_time_color(pos_along_traj, radial_dist)
```
